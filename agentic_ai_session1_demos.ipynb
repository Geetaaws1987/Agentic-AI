## 1 · Traditional programming vs Machine learning
<span style="display:inline-block;background:#f26b1d;color:#fff;font-weight:700;font-size:12px;letter-spacing:.06em;padding:3px 10px;border-radius:999px">SLIDES 5–7</span>

### Traditional programming
The cab company tells us its pricing rule: **₹50 + ₹12/km + ₹2/min**. How do we calculate a fare?

````python
# TRADITIONAL PROGRAMMING: a human writes the rule
def cab_fare(km, minutes):
    return 50 + 12 * km + 2 * minutes      # We KNOW the pricing formula

print(f"Fare for 6 km, 18 min:  ₹{cab_fare(6, 18)}")
````

### Machine learning
Now suppose we **don't know** the rule. We only have 5 past rides and their fares.
Can a machine work out the rule by itself?

````python
from sklearn.linear_model import LinearRegression

# Training data: 5 labelled examples (inputs + correct answer)
#        km  min        fare
X = [[ 5, 15],   #  →  ₹140
     [10, 30],   #  →  ₹230
     [ 2, 10],   #  →  ₹94
     [ 8, 20],   #  →  ₹186
     [12, 25]]   #  →  ₹244

y = [140, 230, 94, 186, 244]   # the labels (correct fares)

model = LinearRegression().fit(X, y)                  # TRAINING
base, per_km, per_min = model.intercept_, *model.coef_
print(f"Learned formula:  Fare = ₹{base:.0f} + ₹{per_km:.0f} per km + ₹{per_min:.0f} per min")

fare = model.predict([[6, 18]])[0]                    # INFERENCE: new input, no label
print(f"Fare for 6 km, 18 min:  ₹{fare:.0f}")
````

> **Takeaway:** Traditional: Rules + Data → Answers. Machine learning: Data + Answers → Rules.

## 2 · Can we write rules for language?
<span style="display:inline-block;background:#f26b1d;color:#fff;font-weight:700;font-size:12px;letter-spacing:.06em;padding:3px 10px;border-radius:999px">SLIDES 8–12</span>

Let's try to detect whether a review is positive or negative using simple word rules.

````python
def sentiment(review):
    review = review.lower()
    if any(w in review for w in ["good", "great", "love", "excellent"]): return "Positive"
    if any(w in review for w in ["bad", "worst", "hate", "terrible"]):   return "Negative"
    return "Neutral"

reviews = ["The food was great",
           "The food was not great",
           "Service was not bad at all",
           "Wow, great, another 2 hour delay"]
for r in reviews: print(f"{sentiment(r):<9} ← {r}")
````

> **Takeaway:** Rules look at **words**, not **meaning**. Negation, double negatives and sarcasm break them.

### GPT-2: a probability for every possible next word

````python
from transformers import AutoTokenizer, AutoModelForCausalLM
tok = AutoTokenizer.from_pretrained("gpt2"); lm = AutoModelForCausalLM.from_pretrained("gpt2")

def next_word(text, min_prob=0.01):
    probs = lm(**tok(text, return_tensors="pt")).logits[0, -1].softmax(-1)
    print(f"\n{text} ___")
    print(f"  It scores all {len(probs):,} possible next words (total = {probs.sum():.0%})")
    for p, i in zip(*probs.sort(descending=True)):
        if p < min_prob: break
        print(f"  {tok.decode(i)!r:14} {p:6.1%}  {'█' * max(1, round(p.item() * 50))}")
    print(f"  ...and {(probs < min_prob).sum():,} more below {min_prob:.0%}")

next_word("It is very cold outside, so wear a")
next_word("The capital of India is New")
next_word("2 + 2 =")
next_word("Please transfer the money to my bank")
next_word("The Spark job crashed with an OutOfMemory")
next_word("She opened the fridge and took out a")
next_word("The latest iPhone model is the iPhone")
next_word("The ball went over the boundary for a")
````

> **Takeaway:** A language model doesn't *know* answers. It predicts what is **likely** to come next.

## 6 · Temperature
<span style="display:inline-block;background:#f26b1d;color:#fff;font-weight:700;font-size:12px;letter-spacing:.06em;padding:3px 10px;border-radius:999px">SLIDES 23–24</span>
<br><span style='color:#6b7385'>The model gives probabilities. To write, it has to pick one word.</span>

### Picking 10 times at different temperatures

````python
import torch

def sample(text, temperature, n=10):
    scores = lm(**tok(text, return_tensors="pt")).logits[0, -1]
    if temperature == 0:
        picks = [scores.argmax()] * n                                   # always the top word
    else:
        probs = (scores / temperature).softmax(-1)                      # temperature reshapes the probabilities
        picks = torch.multinomial(probs, n, replacement=True)           # weighted random pick
    return [tok.decode(i).strip() for i in picks]

examples = ["It is very cold outside, so wear a", "The capital of India is New", "2 + 2 =",
            "Please transfer the money to my bank", "The Spark job crashed with an OutOfMemory",
            "She opened the fridge and took out a", "The latest iPhone model is the iPhone",
            "The ball went over the boundary for a"]

for text in examples:
    print(f"\n{text} ___")
    for T in [0, 0.7, 1.5]:
        print(f"  temperature {T:<4} → {', '.join(sample(text, T))}")
````

### How temperature reshapes the probabilities

````python
text = "Please transfer the money to my bank"
scores = lm(**tok(text, return_tensors="pt")).logits[0, -1]
top4 = scores.topk(4).indices
temps = [0, 0.5, 0.7, 1, 1.5, 2]

def prob(T, i):
    if T == 0: return 1.0 if i == top4[0] else 0.0          # T=0: all probability on the top word
    return (scores / T).softmax(-1)[i].item()

print(f"{'Word':<12}" + "".join(f"{'T=' + str(T):>9}" for T in temps))
for i in top4:
    print(f"{tok.decode(i).strip():<12}" + "".join(f"{prob(T, i):>9.1%}" for T in temps))
print(f"{'All others':<12}" + "".join(f"{1 - sum(prob(T, i) for i in top4):>9.1%}" for T in temps))
````

> **Takeaway:** Low temperature → predictable. High temperature → creative, then nonsense.

### Does vocabulary size matter?

````python
import tiktoken

text = """Hi everyone, our family trip to Kerala is finally confirmed for December!
We fly from Delhi to Thiruvananthapuram on 20th December at 6:45 AM.
Hotel booked in Munnar for 3 nights, total cost Rs 48,500 for all of us.
Day 2 is a houseboat ride in Alleppey, lunch will be Kerala sadya on banana leaf.
सभी लोग अपना आधार कार्ड और टिकट साथ ज़रूर लाएँ।
எல்லோரும் சரியான நேரத்துக்கு விமான நிலையத்துக்கு வந்துவிடுங்கள்.
Bhai log, packing jaldi kar lena, last time bahut late ho gaya tha!
Please carry an umbrella, it rains a lot in the hills.
Total budget per person: Rs 32,000 including flights and food.
Can't wait to see you all! Regards, Venkataraghavan & Lakshmi"""

tokenizers = {"GPT-2 (2019)": "gpt2", "GPT-4 (2023)": "cl100k_base", "GPT-4o (2024)": "o200k_base"}
for label, name in tokenizers.items():
    enc = tiktoken.get_encoding(name)
    n = len(enc.encode(text))
    print(f"{label:<15} vocab {enc.n_vocab:>7,}  →  {n:>3} tokens  {'█' * (n // 10)}")
````

> **Takeaway:** Same message, 56% fewer tokens: the vocabulary grew 4×, and the token count more than halved.

````python
encs = [tiktoken.get_encoding(n) for n in tokenizers.values()]
print(f"{'Line':<45}{'GPT-2':>7}{'GPT-4':>7}{'GPT-4o':>8}")
for line in text.splitlines():
    print(f"{line.strip()[:42]:<45}" + "".join(f"{len(e.encode(line)):>7}" for e in encs))
````

> **Takeaway:** English was already efficient in 2019. Bigger vocabularies mainly help other languages.

## 8 · One token at a time
<span style="display:inline-block;background:#f26b1d;color:#fff;font-weight:700;font-size:12px;letter-spacing:.06em;padding:3px 10px;border-radius:999px">SLIDES 28–29</span>
<br><span style='color:#6b7385'>Predict one token → add it to the input → predict again.</span>

### Watch GPT-2 write, one token at a time

````python
from transformers import AutoTokenizer, AutoModelForCausalLM
tok = AutoTokenizer.from_pretrained("gpt2"); lm = AutoModelForCausalLM.from_pretrained("gpt2")

def generate(text, steps=8):
    for step in range(1, steps + 1):
        ids = tok(text, return_tensors="pt").input_ids
        p, i = lm(ids).logits[0, -1].softmax(-1).max(-1)          # pick the most likely next token
        word = tok.decode(i)
        print(f"Step {step} | input {ids.shape[1]:>2} tokens → {word!r:<11} {p:>4.0%} | {text}[{word}]")
        text += word                                              # feed it back in
    return text

generate("The best way to learn a new language is")
````

## 10 · Base model vs Chat model
<span style="display:inline-block;background:#f26b1d;color:#fff;font-weight:700;font-size:12px;letter-spacing:.06em;padding:3px 10px;border-radius:999px">SLIDES 32–33</span>

### Same family, same size: base vs instruct

````python
from transformers import pipeline
s = "The best way to learn a new language is"

base = pipeline("text-generation", model="Qwen/Qwen2.5-0.5B")
chat = pipeline("text-generation", model="Qwen/Qwen2.5-0.5B-Instruct")

print("BASE:", base(s, max_new_tokens=25, do_sample=False)[0]["generated_text"])
print("\nCHAT:", chat([{"role": "user", "content": s}], max_new_tokens=40, do_sample=False)[0]["generated_text"][-1]["content"])
````

> **Takeaway:** The base model continues your sentence. The chat model treats it as a question and replies. That's instruction tuning.

### Watch the base model write, step by step

````python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
tok = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-0.5B"); lm = AutoModelForCausalLM.from_pretrained("Qwen/Qwen2.5-0.5B")

def generate(text, steps=6, sample=False):
    for step in range(1, steps + 1):
        probs = lm(**tok(text, return_tensors="pt")).logits[0, -1].softmax(-1)   # probability of every token
        top_p, top_i = probs.topk(5)                                              # the 5 most likely
        pick = torch.multinomial(probs, 1)[0] if sample else top_i[0]            # greedy or random pick
        print(f"\nStep {step}: {text} ___")
        for p, i in zip(top_p, top_i):
            print(f"   {tok.decode(i)!r:<12} {p:5.1%}  {'█' * round(p.item() * 40)}{'  ← picked' if i == pick else ''}")
        if pick not in top_i: print(f"   {tok.decode(pick)!r:<12} {probs[pick]:5.1%}  ← picked (outside top 5)")
        text += tok.decode(pick)                                                  # carry it forward
    print(f"\nFinal: {text}")

generate("The best way to learn a new language is")
````

````python
generate("The best way to learn a new language is", sample=True)
````

## 11 · Temperature, one more look
<span style="display:inline-block;background:#f26b1d;color:#fff;font-weight:700;font-size:12px;letter-spacing:.06em;padding:3px 10px;border-radius:999px">SLIDES 34</span>

### A small model (Qwen 0.5B) at different temperatures

````python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
tok = AutoTokenizer.from_pretrained("Qwen/Qwen2.5-0.5B"); lm = AutoModelForCausalLM.from_pretrained("Qwen/Qwen2.5-0.5B")

text = "I opened the fridge and took out a bottle of"
scores = lm(**tok(text, return_tensors="pt")).logits[0, -1]
top = scores.topk(10).indices
temps = [0, 0.5, 1.0, 2.0]

def probability(T):
    if T == 0:
        p = torch.zeros_like(scores); p[top[0]] = 1.0; return p
    return (scores / T).softmax(-1)
probs = {T: probability(T) for T in temps}

print(f"Model: Qwen2.5-0.5B (base)\n")
print(f"{text} ___\n")
print(f"{'Word':<12}{'Score':>7}" + "".join(f"{'T=' + str(T):>9}" for T in temps))
for i in top:
    print(f"{tok.decode(i).strip():<12}{scores[i]:>7.1f}" + "".join(f"{probs[T][i]:>9.1%}" for T in temps))
print(f"{'All others':<12}{'':>7}" + "".join(f"{1 - probs[T][top].sum():>9.1%}" for T in temps))

print("\n10 answers at each temperature:")
for T in temps:
    answers = [tok.decode(i).strip() for i in torch.multinomial(probs[T], 10, replacement=True)]
    print(f"  T={T:<4} {', '.join(answers)}")
````

> **Takeaway:** The scores never change. Temperature decides how much probability stays with the top words and how much leaks out.

### Picking 10 times

````python
from collections import Counter
text = "I opened the fridge and took out a bottle of"

for T in [0, 0.5, 1.0, 2.0]:
    answers = []
    for _ in range(10):
        r = client.chat.completions.create(
            model="openai/gpt-oss-120b", temperature=T, reasoning_effort="low", max_tokens=300,
            messages=[{"role": "system", "content": "Reply with only the single next word that continues the user's text. No punctuation."},
                      {"role": "user", "content": text}])
        answers.append(r.choices[0].message.content.strip().lower())
    print(f"T={T:<4} {', '.join(answers)}")
    print(f"       summary: {dict(Counter(answers))}\n")
````

> **Takeaway:** Small model at temperature 2: nonsense. Big model at temperature 2: still sensible. Bigger models are more confident.

## 12 · Small vs big models
<span style="display:inline-block;background:#f26b1d;color:#fff;font-weight:700;font-size:12px;letter-spacing:.06em;padding:3px 10px;border-radius:999px">SLIDES 35–39</span>

### Same questions: 0.5B vs 120B

````python
import textwrap
from transformers import pipeline
from groq import Groq
from google.colab import userdata

# The two models we're comparing
SMALL_MODEL = "Qwen/Qwen2.5-0.5B-Instruct"     # 0.5 billion parameters, runs on Colab
LARGE_MODEL = "openai/gpt-oss-120b"            # 120 billion parameters, runs on Groq

small = pipeline("text-generation", model=SMALL_MODEL)
client = Groq(api_key=userdata.get("GROQ_API_KEY"))

def ask_small(q):
    out = small([{"role": "user", "content": q}], max_new_tokens=80, do_sample=False)
    return out[0]["generated_text"][-1]["content"].strip()

def ask_large(q):
    r = client.chat.completions.create(model=LARGE_MODEL, reasoning_effort="low",
                                       messages=[{"role": "user", "content": q}])
    return r.choices[0].message.content.strip()

def show(label, answer):
    print(f"  {label}")
    print(textwrap.indent(textwrap.fill(answer, 90), "      "), "\n")

questions = [
    "Which is heavier: a kilogram of cotton or a kilogram of iron? Answer in one sentence.",
    "A shop sells samosas at Rs 15 each. I buy 3 plates with 4 samosas each and pay with a Rs 500 note. "
    "How much change do I get? Give just the number.",
]

for q in questions:
    print("=" * 95)
    print(textwrap.fill(f"Q: {q}", 95), "\n")
    show(f"Small model ({SMALL_MODEL}):", ask_small(q))
    show(f"Large model ({LARGE_MODEL}):", ask_large(q))
````

> **Takeaway:** Choosing a model = balancing **quality, speed and cost**.

## 13 · Reasoning models
<span style="display:inline-block;background:#f26b1d;color:#fff;font-weight:700;font-size:12px;letter-spacing:.06em;padding:3px 10px;border-radius:999px">SLIDES 40–41</span>

### Which models does Groq offer?

### Language model vs reasoning model

````python
import time

def ask(model, question):
    start = time.time()
    r = client.chat.completions.create(model=model, temperature=0, max_tokens=800,
                                       messages=[{"role": "user", "content": question}])
    msg = r.choices[0].message
    return msg.content, getattr(msg, "reasoning", None), time.time() - start, r.usage.completion_tokens

questions = [
    "I am a girl. I have 3 brothers. Each brother has 2 sisters. How many sisters do I have? Answer with just the number.",
    "How many times does the letter 'e' appear in the phrase 'engineering excellence'? Answer with just the number.",
    "A train leaves at 9:40 AM. The journey takes 2 hours 35 minutes, plus a 25-minute halt on the way. When does it arrive? Answer with just the time.",
]
models = [("Language model  (Allam 7B)    ", "allam-2-7b"),
          ("Reasoning model (gpt-oss-120b)", "openai/gpt-oss-120b")]

for q in questions:
    print("QUESTION:", q, "\n")
    for label, model in models:
        answer, thinking, secs, tokens = ask(model, q)
        print(f"{label}: {answer.strip()}   ({secs:.1f}s, {tokens} output tokens)")
        if thinking:
            print(f"   Its thinking: {thinking.strip()[:500]}...")
    print("-" * 90)
````

> **Takeaway:** Reasoning models think before answering: better on logic and maths, but slower and you pay for the thinking tokens.

## 14 · Speech models: a voice assistant
<span style="display:inline-block;background:#f26b1d;color:#fff;font-weight:700;font-size:12px;letter-spacing:.06em;padding:3px 10px;border-radius:999px">SLIDES 44–46</span>
<br><span style='color:#6b7385'>Your voice → Whisper (hear) → gpt-oss-120b (think) → Orpheus (speak)</span>

### Record your question

````python
from IPython.display import Javascript, display, Audio
from google.colab.output import eval_js
from base64 import b64decode

def record(seconds=5):
    display(Javascript("""
    async function rec(ms) {
      const s = await navigator.mediaDevices.getUserMedia({audio: true});
      const r = new MediaRecorder(s), chunks = [];
      r.ondataavailable = e => chunks.push(e.data); r.start();
      await new Promise(x => setTimeout(x, ms)); r.stop();
      await new Promise(x => r.onstop = x); s.getTracks().forEach(t => t.stop());
      const buf = new Uint8Array(await new Blob(chunks).arrayBuffer());
      let bin = ""; buf.forEach(v => bin += String.fromCharCode(v)); return btoa(bin);
    }"""))
    print(f"🎤 Speak now ({seconds} seconds)...")
    open("question.webm", "wb").write(b64decode(eval_js(f"rec({seconds * 1000})")))
    print("✅ Recorded")
    return "question.webm"
````

### Speech → text (Whisper) → answer (gpt-oss-120b)

````python
audio_file = record(5)

# Step 1: speech → text (Whisper)
with open(audio_file, "rb") as f:
    question = client.audio.transcriptions.create(file=(audio_file, f.read()), model="whisper-large-v3").text
print("You said:", question)

# Step 2: text → answer (language model)
answer = client.chat.completions.create(
    model="openai/gpt-oss-120b", reasoning_effort="low", max_tokens=500,
    messages=[{"role": "system", "content": "Answer in 2-3 short sentences, in the same language as the question."},
              {"role": "user", "content": question}]).choices[0].message.content
print("Answer:", answer)
````

### Text → speech (Orpheus)

````python
from IPython.display import Audio

speech = client.audio.speech.create(
    model="canopylabs/orpheus-v1-english",
    voice="troy",                     # other voices in Groq's docs: "hannah", "austin"
    input=answer,
    response_format="wav")
speech.write_to_file("answer.wav")
Audio("answer.wav", autoplay=True)
````

> **Takeaway:** Real applications chain models together: the first step towards an agent.

<br><span style='color:#6b7385'>Same 120B model, simple questions.</span>

### Setup

````python
!pip -q install groq
from groq import Groq
from google.colab import userdata

client = Groq(api_key=userdata.get("GROQ_API_KEY"))
MODEL = "openai/gpt-oss-120b"

def ask(question, history=None):
    messages = (history or []) + [{"role": "user", "content": question}]
    r = client.chat.completions.create(model=MODEL, messages=messages, max_tokens=1500, reasoning_effort="low")
    return r.choices[0].message.content.strip()
````

### 1. Knowledge cutoff

````python
print(ask("What is your knowledge cutoff date?"))
print(ask("Who won IPL 2026?"))
````

### 2. No live information

````python
print(ask("What is the weather in Bengaluru right now? And what is the current price of gold in India?"))
````

### 3. No memory

````python
print(ask("Hi, my name is Sumit & I teach Agentic AI"))
print(ask("What is my name and what do I teach?"))
````

### 4. Bad at exact maths
A 120B reasoning model can often work out small sums step by step, so we use a harder calculation.

````python
import time
q = "What is 4829173 × 9364821? Answer with just the number."
start = time.time()
print("Model: ", ask(q), f"  ({time.time() - start:.1f}s)")
print("Python:", 4829173 * 9364821, "  (instant)")
````
